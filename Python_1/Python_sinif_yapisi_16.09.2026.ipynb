```python
# Kendi çalışma notu: Sınıf örnekleri için nesne durumunu ayrıca gözden geçir.
class Araba:
    def __init__(self, marka, model, hiz_kms=0):
        self.marka = marka
        self.model = model
        self.hiz_kms = hiz_kms

    def hizlandir(self, add):
        self.hiz_kms += add

    def bilgi_goster(self):
        print(f"{self.marka} {self.model} - Hız: {self.hiz_kms} km/s")


arac = Araba("VW", "Scirocco", 170)
arac.hizlandir(40)
arac.bilgi_goster()
```

```python
class Dikdortgen:
    def __init__(self, uzunluk, genislik):
        self.genislik = genislik;
        self.uzunluk = uzunluk;
        

    def alan(self):
        return self.uzunluk * self.genislik

    def cevre(self):
        return 2 * (self.uzunluk + self.genislik)


dikdortgen = Dikdortgen(10, 5)

print("Alan: ", dikdortgen.alan())
print("Çevre: ", dikdortgen.cevre())
```

```python
class Kitap:
    def __init__(self, ad, yazar, sayfa):
        self.ad = ad;
        self.sayfa = sayfa;
        self.yazar = yazar;

    def ozet(self):
        return f"{self.ad} - {self.yazar} ({self.sayfa} sayfa)"


kitap = Kitap("Dil Belası", "İmam Gazali rh.", 123)
print(kitap.ozet())
```

### Çalışma değerlendirmesi

Sınıf örneklerinde nesnenin durumunu metotlarla okumak, aynı veriyi farklı örneklerde tekrar kullanmayı kolaylaştırır.


### Ek uygulama: sınıf davranışını doğrulama

**Amaç:** Kurucu metotla saklanan verinin metot sonuçlarına doğru yansıdığını sınamak.

**Kontrol:** Dikdörtgenin alanı ve çevresi beklenen değerleri vermeli; kitap özeti başlık, yazar ve sayfa sayısını içermeli. Bu küçük kontroller sınıfları değiştirirken önceki davranışın bozulmasını fark etmeyi sağlar.

```python
assert dikdortgen.alan() == 50
assert dikdortgen.cevre() == 30
assert "Dil Belası" in kitap.ozet()
print("Sınıf davranışı kontrolleri başarılı.")
```
